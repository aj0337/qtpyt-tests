```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
E, T_dft = np.load("./output/lowdin/dft/ET.npy")
E, T_ed = np.load("./output/lowdin/ed/ET.npy")
```

```python


plt.style.use("dark_background")

fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(
    E,
    T_dft,
    lw=3.5,
    color="#00D4FF",
    label="DFT"
)

ax.plot(
    E,
    T_ed,
    lw=3.5,
    color="#FFB000",
    label="ED"
)

ax.set_yscale("log")

ax.set_xlabel(
    "Energy (eV)",
    fontsize=18,
    fontweight="bold"
)

ax.set_ylabel(
    "Transmission",
    fontsize=18,
    fontweight="bold"
)

ax.set_xlim(-3, 3)
ax.set_ylim(bottom=1e-5)

ax.tick_params(
    axis="both",
    which="major",
    labelsize=14,
    width=1.5,
    length=6
)

ax.grid(
    alpha=0.25,
    linestyle="--"
)

ax.legend(
    fontsize=15,
    frameon=False,
    loc="best"
)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()

plt.savefig(
    "transmission_social.png",
    dpi=600,
    bbox_inches="tight"
)

plt.show()
```

```python
plt.style.use("dark_background")

fig, ax = plt.subplots(figsize=(12, 7))

ax.plot(E, T_dft, lw=4.0, color="#00D4FF", label="DFT")
ax.plot(E, T_ed, lw=4.0, color="#FFB000", label="ED")

# ax.axvline(0, color="white", lw=2.0, alpha=0.45, linestyle=":")

ax.set_yscale("log")
ax.set_xlabel("Energy (eV)", fontsize=20, fontweight="bold")
ax.set_ylabel("Transmission", fontsize=20, fontweight="bold")

ax.set_xlim(-3, 3)
ax.set_ylim(bottom=1e-5)

ax.tick_params(axis="both", which="major", labelsize=16, width=1.8, length=7)
ax.grid(alpha=0.22, linestyle="--")

ax.legend(fontsize=17, frameon=False, loc="best")

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
arrow_y = 1e-1
x_start = -0.8
x_end = 1.4

ax.annotate(
    "",
    xy=(x_start, arrow_y),
    xytext=(x_end, arrow_y),
    arrowprops=dict(
        arrowstyle="<->",
        lw=2.5,
        color="white",
    ),
)

ax.text(
    0.2 * (x_start + x_end),
    8e-2,
    "Splitting",
    ha="center",
    va="top",
    fontsize=16,
    fontweight="bold",
    color="white",
)
plt.tight_layout()
plt.savefig("transmission_linkedin.png", dpi=600, bbox_inches="tight")
plt.show()
```
